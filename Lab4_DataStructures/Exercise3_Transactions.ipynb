```python
from collections import defaultdict

# STUDENT-SPECIFIC INPUTS
LAST_NAME = "Correa"
SEED_NUM = 6
FAVORITE_ARTIST = "Jennie"

# Convert artist's letters to numerical positions (A=1, B=2, ..., Z=26)
artist_values = [
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST if char.isalpha()
]

print("STUDENT INPUTS")
print(f"Last Name: {LAST_NAME}")
print(f"Seed Number: {SEED_NUM}")
print(f"Favorite Artist: {FAVORITE_ARTIST}")
print(f"Character Values: {artist_values}")


# 1. GENERATE TRANSACTION RECORDS
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

products = [
    "NOTEBOOK",
    "PEN",
    "USB",
    "HEADPHONES",
    "CALCULATOR"
]

# Fixed example prices for each product
prices = {
    "NOTEBOOK": 35.00,
    "PEN": 15.00,
    "USB": 250.00,
    "HEADPHONES": 350.00,
    "CALCULATOR": 180.00
}

transaction_records = []

# Generate transactions for each customer and product
for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        char_value = artist_values[(i + j) % len(artist_values)]

        quantity = 1 + (
            (SEED_NUM + char_value + (i * 3) + (j * 2)) % 5
        )

        transaction_records.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "price": prices[product]
        })

# Add one repeated customer-product purchase
transaction_records.append({
    "customer": customers[0],
    "product": products[0],
    "quantity": 2,
    "price": prices[products[0]]
})

print("\n1. GENERATED TRANSACTION RECORDS")
for record in transaction_records:
    print(record)


# 2. ORGANIZE TRANSACTION DATA
organized_records = sorted(
    transaction_records,
    key=lambda record: (
        record["customer"],
        record["product"]
    )
)

print("\n2. ORGANIZED TRANSACTION DATA")
for record in organized_records:
    print(record)


# 3. IDENTIFY REPEATED AND DISTINCT TRANSACTIONS
purchase_counts = defaultdict(int)

for record in transaction_records:
    key = (record["customer"], record["product"])
    purchase_counts[key] += 1

repeated_purchases = [
    (key, count)
    for key, count in sorted(purchase_counts.items())
    if count > 1
]

distinct_customers = sorted({
    record["customer"] for record in transaction_records
})

distinct_products = sorted({
    record["product"] for record in transaction_records
})

print("\n3. REPEATED / DISTINCT TRANSACTIONS")

print("\nRepeated Customer-Product Purchases:")
if repeated_purchases:
    for key, count in repeated_purchases:
        print(key, "appears", count, "times")
else:
    print("No repeated purchases.")

print("\nDistinct Customers:", distinct_customers)
print("Number of Distinct Customers:", len(distinct_customers))

print("\nDistinct Products:", distinct_products)
print("Number of Distinct Products:", len(distinct_products))


# 4. CALCULATE TRANSACTION VALUES
for record in transaction_records:
    record["transaction_value"] = (
        record["quantity"] * record["price"]
    )

total_transaction_value = sum(
    record["transaction_value"]
    for record in transaction_records
)

print("\n4. TRANSACTION VALUES")
for record in transaction_records:
    print(
        record["customer"],
        "|", record["product"],
        "| Quantity:", record["quantity"],
        "| Price: ₱", f"{record['price']:.2f}",
        "| Transaction Value: ₱",
        f"{record['transaction_value']:.2f}"
    )

print(
    "\nOverall Transaction Value: ₱",
    f"{total_transaction_value:.2f}"
)


# 5. CUSTOMER SUMMARIES
customer_totals = defaultdict(float)
customer_transaction_counts = defaultdict(int)
customer_quantities = defaultdict(int)

for record in transaction_records:
    customer = record["customer"]
    customer_totals[customer] += record["transaction_value"]
    customer_transaction_counts[customer] += 1
    customer_quantities[customer] += record["quantity"]

print("\n5. CUSTOMER SUMMARIES")
for customer in customers:
    print({
        "customer": customer,
        "number_of_transactions": customer_transaction_counts[customer],
        "total_quantity": customer_quantities[customer],
        "total_spending": round(customer_totals[customer], 2)
    })


# 6. PRODUCT SUMMARIES
product_totals = defaultdict(float)
product_quantities = defaultdict(int)
product_transaction_counts = defaultdict(int)

for record in transaction_records:
    product = record["product"]
    product_totals[product] += record["transaction_value"]
    product_quantities[product] += record["quantity"]
    product_transaction_counts[product] += 1

print("\n6. PRODUCT SUMMARIES")
for product in products:
    print({
        "product": product,
        "price": prices[product],
        "number_of_transactions": product_transaction_counts[product],
        "total_quantity_sold": product_quantities[product],
        "total_sales": round(product_totals[product], 2)
    })


# 7. ORDERED RESULTS
print("\n7. ORDERED RESULTS")

print("\nTransactions ordered by customer and product:")
for record in organized_records:
    print({
        **record,
        "transaction_value": round(
            record["quantity"] * record["price"], 2
        )
    })

print("\nTransactions ordered from highest to lowest value:")
ranked_records = sorted(
    transaction_records,
    key=lambda record: record["transaction_value"],
    reverse=True
)

for record in ranked_records:
    print(
        record["customer"],
        "|", record["product"],
        "| Value: ₱",
        f"{record['transaction_value']:.2f}"
    )


# 8. OVERALL TRANSACTION SUMMARY
total_transactions = len(transaction_records)
total_units_sold = sum(
    record["quantity"] for record in transaction_records
)

average_transaction_value = (
    total_transaction_value / total_transactions
    if total_transactions > 0 else 0
)

highest_transaction = max(
    transaction_records,
    key=lambda record: record["transaction_value"]
)

lowest_transaction = min(
    transaction_records,
    key=lambda record: record["transaction_value"]
)

print("\n8. OVERALL TRANSACTION SUMMARY")
print("Total Transaction Records:", total_transactions)
print("Distinct Customers:", len(distinct_customers))
print("Distinct Products:", len(distinct_products))
print("Repeated Customer-Product Pairs:", len(repeated_purchases))
print("Total Units Sold:", total_units_sold)
print("Overall Transaction Value: ₱", f"{total_transaction_value:.2f}")
print("Average Transaction Value: ₱", f"{average_transaction_value:.2f}")

print(
    "Highest Transaction:",
    highest_transaction["customer"],
    "|", highest_transaction["product"],
    "| ₱", f"{highest_transaction['transaction_value']:.2f}"
)

print(
    "Lowest Transaction:",
    lowest_transaction["customer"],
    "|", lowest_transaction["product"],
    "| ₱", f"{lowest_transaction['transaction_value']:.2f}"
)

print("\nFINAL CONSOLIDATED TRANSACTION REPORT")

print("\nCustomer Spending:")
for customer in customers:
    print(customer, "-> ₱", f"{customer_totals[customer]:.2f}")

print("\nProduct Sales:")
for product in products:
    print(product, "-> ₱", f"{product_totals[product]:.2f}")

print("\nTotal Transactions:", total_transactions)
print("Total Units Sold:", total_units_sold)
print("Total Sales: ₱", f"{total_transaction_value:.2f}")

print("\nTransaction history analysis completed successfully.")
```
