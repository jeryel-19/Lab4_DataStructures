```python
from collections import defaultdict

# STUDENT-SPECIFIC INPUTS

LAST_NAME = "Correa"
SEED_NUM = 6
FAVORITE_ARTIST = "Jennie"

# Convert artist's letters to numerical positions (A=1, B=2, ..., Z=26)
artist_values = [
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST if char.isalpha()
]

print(f"STUDENT INPUTS")
print(f"Last Name: {LAST_NAME}")
print(f"Seed Number: {SEED_NUM}")
print(f"Favorite Artist: {FAVORITE_ARTIST}")
print(f"Character Values: {artist_values}")

# 1. GENERATE TEST DATA
sources = [
    f"{LAST_NAME}_T1",
    f"{LAST_NAME}_T2",
    f"{LAST_NAME}_T3",
    f"{LAST_NAME}_T4",
    f"{LAST_NAME}_TS"
]

categories = [
    "TEMPERATURE",
    "VOLTAGE",
    "CURRENT",
    "PRESSURE",
    "SPEED"
]

test_records = []

# Generate five initial records: one for each source/category
for i in range(5):
    value = SEED_NUM + artist_values[i % len(artist_values)] + (i * 2)

    test_records.append({
        "source": sources[i],
        "category": categories[i],
        "value": value
    })

# Add four measurements under different existing source/category pairs
# Each added pair is different from the initial five pairs.
for i in range(4):
    source_index = i
    category_index = (i + 1) % len(categories)
    char_value = artist_values[(i + 5) % len(artist_values)]
    value = SEED_NUM + char_value + ((i + 5) * 2)

    test_records.append({
        "source": sources[source_index],
        "category": categories[category_index],
        "value": value
    })

# Add exactly one extra record repeating an existing source/category pair
test_records.append({
    "source": sources[0],
    "category": categories[0],
    "value": SEED_NUM + artist_values[0] + 25
})

print(f"\n1. GENERATED TEST DATA")
for record in test_records:
    print(record)


# 2. ORGANIZE THE RECORDS
organized = sorted(
    test_records,
    key=lambda record: (
        record["source"],
        record["category"],
        record["value"]
    )
)

print(f"\n2. ORGANIZED TEST DATA")
for record in organized:
    print(record)


# 3. IDENTIFY REPEATED AND DISTINCT DATA
source_category_counts = defaultdict(int)
exact_record_counts = defaultdict(int)

for record in test_records:
    source_category_key = (
        record["source"],
        record["category"]
    )

    exact_record_key = (
        record["source"],
        record["category"],
        record["value"]
    )

    source_category_counts[source_category_key] += 1
    exact_record_counts[exact_record_key] += 1

print(f"\n3. REPEATED / DISTINCT DATA")

print(f"\nRepeated source/category combinations:")
repeated_combinations = [
    (key, count)
    for key, count in sorted(source_category_counts.items())
    if count > 1
]

if repeated_combinations:
    for key, count in repeated_combinations:
        print(key, "appears", count, "times")
else:
    print("No repeated source/category combinations.")

print(f"\nExact duplicate records:")
exact_duplicates = [
    (key, count)
    for key, count in sorted(exact_record_counts.items())
    if count > 1
]

if exact_duplicates:
    for key, count in exact_duplicates:
        print(key, "appears", count, "times")
else:
    print("No exact duplicate records.")

distinct_sources = sorted({
    record["source"] for record in test_records
})

distinct_categories = sorted({
    record["category"] for record in test_records
})

print(f"\nDistinct Sources: {distinct_sources}")
print(f"Number of Distinct Sources: {len(distinct_sources)}")
print(f"Distinct Categories: {distinct_categories}")
print(f"Number of Distinct Categories: {len(distinct_categories)}")
print(f"Number of Distinct Exact Records: {len(exact_record_counts)}")


# 4. NUMERICAL AND CATEGORICAL SUMMARIES
category_values = defaultdict(list)
source_values = defaultdict(list)

for record in test_records:
    category_values[record["category"]].append(record["value"])
    source_values[record["source"]].append(record["value"])

print(f"\n4. NUMERICAL AND CATEGORICAL SUMMARIES")

print(f"\nCategorical Summary: Measurement Count by Category")
for category in sorted(category_values):
    print(category, "->", len(category_values[category]), "measurements")

print(f"\nNumerical Summary by Category")
for category, values in sorted(category_values.items()):
    print({
        "category": category,
        "count": len(values),
        "minimum": min(values),
        "maximum": max(values),
        "total": sum(values),
        "average": round(sum(values) / len(values), 2)
    })

print("\nNumerical Summary by Source")
for source, values in sorted(source_values.items()):
    print({
        "source": source,
        "count": len(values),
        "minimum": min(values),
        "maximum": max(values),
        "total": sum(values),
        "average": round(sum(values) / len(values), 2)
    })

# 5. PROCESSED RESULTS
print("\n5. PROCESSED RESULTS")

for category, values in sorted(category_values.items()):
    print(
        category,
        "| Total =", sum(values),
        "| Measurements =", len(values),
        "| Average =", round(sum(values) / len(values), 2)
    )


# 6. ORDERED RESULTS
print("\n6. ORDERED RESULTS")
print("Records sorted by source, category, and value:")

for record in organized:
    print(record)


# 7. FINAL SUMMARY
all_values = [
    record["value"] for record in test_records
]

print(f"\n7. FINAL ORGANIZED AND SUMMARIZED TEST DATASET")
print(f"Total Measurement Records: {len(test_records)}")
print(f"Distinct Test Sources: {len(distinct_sources)}")
print(f"Distinct Categories: {len(distinct_categories)}")
print(f"Distinct Exact Records: {len(exact_record_counts)}")
print(f"Repeated Source/Category Combinations: {len(repeated_combinations)}")
print(f"Overall Measurement Total: {sum(all_values)}")
print(f"Overall Minimum Value: {min(all_values)}")
print(f"Overall Maximum Value: {max(all_values)}")
print(f"Overall Measurement Average: {round(sum(all_values) / len(all_values), 2)}"
)

print(f"\nFinal Organized Dataset:")
for record in organized:
    print(record)

print(f"\nDataset generation and analysis completed successfully.")
```
