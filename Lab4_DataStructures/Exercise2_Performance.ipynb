```python
from collections import defaultdict

# STUDENT-SPECIFIC INPUTS
LAST_NAME = "Correa"
SEED_NUM = 6
FAVORITE_ARTIST = "Jennie"

# Convert artist's letters to numerical positions (A=1, B=2, ..., Z=26)
artist_values = [
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST if char.isalpha()
]

print("STUDENT INPUTS")
print(f"Last Name: {LAST_NAME}")
print(f"Seed Number: {SEED_NUM}")
print(f"Favorite Artist: {FAVORITE_ARTIST}")
print(f"Character Values: {artist_values}")


# 1. GENERATE STUDENT PERFORMANCE DATA
students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

performance_records = []

# Generate one score for every student and subject
for i, student in enumerate(students):
    for j, subject in enumerate(subjects):
        char_value = artist_values[(i + j) % len(artist_values)]

        score = 50 + (
            (SEED_NUM + char_value + (i * 7) + (j * 5)) % 51
        )

        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score
        })

# Include one missing score in English
for record in performance_records:
    if (
        record["student"] == students[0]
        and record["subject"] == "ENGLISH"
    ):
        record["score"] = None
        break

# Add one exact duplicate record to identify repeated data
performance_records.append(performance_records[0].copy())

print("\n1. GENERATED STUDENT PERFORMANCE DATA")
for record in performance_records:
    print(record)


# 2. ORGANIZE THE RECORDS
organized = sorted(
    performance_records,
    key=lambda record: (
        record["student"],
        record["subject"]
    )
)

print("\n2. ORGANIZED STUDENT PERFORMANCE DATA")
for record in organized:
    print(record)


# 3. IDENTIFY REPEATED, DISTINCT, AND INCOMPLETE DATA
record_counts = defaultdict(int)

for record in performance_records:
    exact_key = (
        record["student"],
        record["subject"],
        record["score"]
    )
    record_counts[exact_key] += 1

print("\n3. REPEATED / DISTINCT / INCOMPLETE DATA")

print("\nRepeated exact records:")
repeated_records = [
    (key, count)
    for key, count in sorted(
        record_counts.items(),
        key=lambda item: (
            item[0][0],
            item[0][1],
            -1 if item[0][2] is None else item[0][2]
        )
    )
    if count > 1
]

if repeated_records:
    for key, count in repeated_records:
        print(key, "appears", count, "times")
else:
    print("No repeated exact records.")

distinct_students = sorted({
    record["student"] for record in performance_records
})

distinct_subjects = sorted({
    record["subject"] for record in performance_records
})

incomplete_records = [
    record for record in performance_records
    if record["score"] is None
]

print(f"\nDistinct Students: {distinct_students}")
print(f"Number of Distinct Students: {len(distinct_students)}")

print(f"\nDistinct Subjects: {distinct_subjects}")
print(f"Number of Distinct Subjects: {len(distinct_subjects)}")

print("\nIncomplete Records:")
for record in incomplete_records:
    print(record)

print("Number of Incomplete Records:", len(incomplete_records))


# 4. NUMERICAL AND CATEGORICAL SUMMARIES
def classify_score(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "GOOD"
    elif score >= 70:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"


# Remove exact duplicate records for calculations
unique_records = list({
    (
        record["student"],
        record["subject"],
        record["score"]
    ): record
    for record in performance_records
}.values())

valid_records = [
    record for record in unique_records
    if record["score"] is not None
]

student_scores = defaultdict(list)
subject_scores = defaultdict(list)
classification_counts = defaultdict(int)

for record in unique_records:
    classification = classify_score(record["score"])
    classification_counts[classification] += 1

    if record["score"] is not None:
        student_scores[record["student"]].append(record["score"])
        subject_scores[record["subject"]].append(record["score"])

print("\n4. NUMERICAL AND CATEGORICAL SUMMARIES")

print("\nCategorical Summary: Score Classifications")
for classification, count in sorted(classification_counts.items()):
    print(classification, "->", count, "records")

print("\nNumerical Summary by Student")
for student in students:
    scores = student_scores[student]

    if scores:
        print({
            "student": student,
            "number_of_scores": len(scores),
            "minimum": min(scores),
            "maximum": max(scores),
            "total": sum(scores),
            "average": round(sum(scores) / len(scores), 2)
        })
    else:
        print(student, "has no valid scores.")

print("\nNumerical Summary by Subject")
for subject, scores in sorted(subject_scores.items()):
    print({
        "subject": subject,
        "number_of_scores": len(scores),
        "minimum": min(scores),
        "maximum": max(scores),
        "total": sum(scores),
        "average": round(sum(scores) / len(scores), 2)
    })


# 5. PROCESSED RESULTS
print("\n5. PROCESSED RESULTS")

print("\nStudent Performance Results")
for student in students:
    scores = student_scores[student]

    if scores:
        print(
            student,
            "| Valid Scores =", len(scores),
            "| Total =", sum(scores),
            "| Average =", round(sum(scores) / len(scores), 2)
        )

print("\nSubject Performance Results")
for subject, scores in sorted(subject_scores.items()):
    print(
        subject,
        "| Valid Scores =", len(scores),
        "| Total =", sum(scores),
        "| Average =", round(sum(scores) / len(scores), 2)
    )


# 6. ORDERED RESULTS
print("\n6. ORDERED RESULTS")

print("\nRecords ordered by student and subject:")
for record in sorted(
    unique_records,
    key=lambda record: (
        record["student"],
        record["subject"]
    )
):
    result = record.copy()
    result["classification"] = classify_score(record["score"])
    print(result)

print("\nScores ordered from highest to lowest:")
ranked_records = sorted(
    valid_records,
    key=lambda record: record["score"],
    reverse=True
)

for record in ranked_records:
    print(
        record["student"],
        "|", record["subject"],
        "| Score:", record["score"],
        "|", classify_score(record["score"])
    )


# 7. FINAL SUMMARY
all_scores = [
    record["score"] for record in unique_records
    if record["score"] is not None
]

print("\n7. FINAL ORGANIZED AND SUMMARIZED DATASET")
print("Total Performance Records:", len(performance_records))
print("Unique Performance Records:", len(unique_records))
print("Distinct Students:", len(distinct_students))
print("Distinct Subjects:", len(distinct_subjects))
print("Valid Scores:", len(all_scores))
print(
    "Missing Scores:",
    sum(1 for record in unique_records if record["score"] is None)
)
print("Repeated Exact Record Types:", len(repeated_records))
print("Total of Valid Scores:", sum(all_scores))
print("Overall Minimum Score:", min(all_scores))
print("Overall Maximum Score:", max(all_scores))
print(
    "Overall Average Score:",
    round(sum(all_scores) / len(all_scores), 2)
)

print("\nFinal Organized Dataset:")
for record in sorted(
    unique_records,
    key=lambda record: (
        record["student"],
        record["subject"]
    )
):
    result = record.copy()
    result["classification"] = classify_score(record["score"])
    print(result)

print("\nStudent performance data analysis completed successfully.")
```
